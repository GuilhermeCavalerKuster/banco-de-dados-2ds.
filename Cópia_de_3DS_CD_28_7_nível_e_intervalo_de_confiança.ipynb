```python
# Importa bibliotecas necessárias do Python
import numpy as np  # Para operações numéricas (cálculo de médias e desvios padrão).
import scipy.stats as stats # Para funções estatísticas (cálculo do Z-score).
```

```python
# Dados de exemplo: horas de estudo por semana (amostra)
horas_estudo = [15, 18, 12, 20, 21, 22, 14, 19, 13, 17, 15, 18, 16, 19, 20, 17, 16, 14, 15, 18,
                19, 13, 20, 21, 22, 14, 15, 17, 18, 19, 16, 17, 14, 15, 18, 19, 13, 20, 21,
                22, 14, 15, 17, 18, 19, 16, 17, 14, 15, 18]

# Média e desvio padrão da amostra
media_amostral = np.mean(horas_estudo)
desvio_padrao_amostral = np.std(horas_estudo, ddof=1)
tamanho_amostra = len(horas_estudo)

# Se estivesse calculando o desvio padrão de todos os alunos da escola (população),
# usaria o padrão do Python. Como usa-se só uma amostra de 50 alunos para "dar um
# palpite" sobre o total, usa-se ddof=1 para o cálculo ser + preciso/seguro contra subestimações.

# Exibe valores calculados na tela
print(f"Média Amostral: {media_amostral} hs")
print(f"Desvio Padrão Amostral: {desvio_padrao_amostral} hs de estudo/semana")
print(f"Tamanho da Amostra: {tamanho_amostra} medições")
```

```python
# Nível de confiança
nivel_confianca = 0.95
z = stats.norm.ppf(1 - (1 - nivel_confianca) / 2)

# Calcular o erro padrão da média
erro_padrao = desvio_padrao_amostral / np.sqrt(tamanho_amostra)

# Exibe o nível e confiança e erro padrão da média na tela
print(f"Nível de Confiança: {nivel_confianca} ou {nivel_confianca * 100}%")
print(f"Z-score: {z} (ou multiplicador que dá o intervalo de variação)")
print(f"Erro Padrão da Média: {erro_padrao} hs de estudo/semana")
```

```python
# Calcular o intervalo de confiança
margem_erro = z * erro_padrao
intervalo_confianca = (media_amostral - margem_erro, media_amostral + margem_erro)

# Exibe a média amostral e intervalo de confiança na tela
print(f"Média amostral: {media_amostral:.2f} hs de estudo/semana")
print(f"Intervalo de confiança de {nivel_confianca * 100}%: entre {intervalo_confianca[0]:.2f} e {intervalo_confianca[1]:.2f} hs")
```
